```python
#If elif else statement fro marks of the student
# Above 95 - grade A , 85 to 95 - grade B , 75 to 85 - grade C , 65 to 75 - grade D , below 65 - grade F 
marks = int(input("Enter your marks: "))
if marks > 95:
    print("Grade A")
elif marks >= 85:
    print("Grade B")
elif marks >= 75:
    print("Grade C")
elif marks >= 65:
    print("Grade D")
else:
    print("Grade F")    
```
